### 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

Llista de publicacions (/posts)
Llista d’usuaris (/users)
Llista de tasques (/todos)

### 2. Mostra per pantalla:

La quantitat total de cada recurs.
El codi d'estat de cada petició.


```python
recursos = ["posts", "users", "todos"]

for recurs in recursos:
    # Fem la petició GET per obtenir la representació del recurs [4, 5]
    resposta = requests.get(f"{base_url}/{recurs}")
    
    # Obtenim les dades en format llista/diccionari de Python [6, 7]
    dades = resposta.json()
    
    # Mostrem la quantitat total i el codi d'estat (200 OK si ha anat bé) [8, 9]
    print(f"Recurs: /{recurs}")
    print(f"  - Quantitat total: {len(dades)}")
    print(f"  - Codi d'estat: {resposta.status_code}")
    print("-" * 30)
```

### 3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
# Intentem accedir a una publicació amb un ID molt alt que no existeix [10, 11]
resposta_inexistent = requests.get(f"{base_url}/posts/99999")
print(f"Petició a /posts/99999 - Codi d'estat: {resposta_inexistent.status_code} (Not Found)")
print("-" * 30)
```

### 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
nova_publicacio = {
    "title": "Això és un títol de prova",
    "body": "Contingut del missatge creat des de Python.",
    "userId": 1
}

# Enviem les dades utilitzant el paràmetre 'json' per a la serialització automàtica [12, 13]
resposta_post = requests.post(f"{base_url}/posts", json=nova_publicacio)

print("Resposta POST (Creació):")
print(f"  - Resposta JSON: {resposta_post.json()}")
print(f"  - Codi d'estat: {resposta_post.status_code} (Created)") # 201 Created [8]
print("-" * 30)
```

### 5. Fes una petició PATCH per modificar parcialment una publicació existent.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
# Modifiquem només el títol de la publicació amb ID 1 [14]
dades_parcials = {"title": "Títol modificat amb PATCH"}
resposta_patch = requests.patch(f"{base_url}/posts/1", json=dades_parcials)

print("Resposta PATCH (Modificació parcial):")
print(f"  - Resposta JSON: {resposta_patch.json()}")
print(f"  - Codi d'estat: {resposta_patch.status_code}")
print("-" * 30)
```


### 6. Fes una petició DELETE sobre una publicació.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python

# Eliminem la publicació amb ID 1 [12, 14]
resposta_delete = requests.delete(f"{base_url}/posts/1")

print("Resposta DELETE (Eliminació):")
print(f"  - Resposta JSON (normalment buida): {resposta_delete.json()}")
print(f"  - Codi d'estat: {resposta_delete.status_code}") # 200 OK o 204 No Content [8]
```
