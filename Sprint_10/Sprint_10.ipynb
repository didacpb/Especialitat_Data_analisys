```python
# Crea una columna que sigui el nom complet.
df_cap['Nom complet'] = df_cap['Nom'] + " " + df_cap['Cognoms']

# Crea una columna si la persona és nascuda a Espanya o no.
df_cap['Naixament_Espanya'] = np.where(df_cap["País d'origen"] == 'Espanya', 'Sí', 'No')

# Posa el DNI com a índex del DataFrame (noms de files).
df_cap=df_cap.set_index('DNI')
```

```python
# Substitueix el nom de les columnes Dia de Naixement, Mes de Naixement i Any de Naixement per Dia, Mes i Any.
df_cap = df_cap.rename(columns={'Dia de Naixement': 'Dia', 'Mes de Naixement': 'Mes', 'Any de Naixement':'Any'})
# Substitueix H per Home, D per Dona, A per Altres i NC per una dada faltant (nan/null/na).
df_cap['Gènere'] = df_cap['Gènere'].replace('H', 'Home')
df_cap['Gènere'] = df_cap['Gènere'].replace('D', 'Dona')
df_cap['Gènere'] = df_cap['Gènere'].replace('A', 'Altres')
df_cap['Gènere'] = df_cap['Gènere'].replace('NC', np.nan )

# Mostra tots els canvis que has realitzat en una sola taula.
df_cap
```

#### 3.
Junta les columnes Fills i No Fills en una sola columna, utilitzant el mètode .apply() i definint una funció que resolgui el problema. La columna nova ha de dir-se "Fills" i prendre els valors "Sí" o "No".

```python
def fills_si_no(x):
        if x == True:
            return'Sí'
        else:
            return 'No'
        

df_cap['Fills']=df_cap['Fills'].apply(fills_si_no)
df_cap.drop('No Fills', axis='columns', inplace=True)

df_cap
```

#### 4.
Crea una taula resum que permeti veure el sou mig, medià, mínim i màxim per Gènere.

Ordena la taula en funció del sou mig.

```python

df_cap['Salari mensual'] = df_cap['Salari mensual'].astype(str)

df_cap['Salari mensual'] = (df_cap['Salari mensual'].str.replace('€', '', regex=False).str.replace('.', '', regex=False).str.strip())

df_cap['Salari mensual'] = pd.to_numeric(df_cap['Salari mensual'], errors='coerce')


df_sous = df_cap.groupby('Gènere')['Salari mensual'].agg(['mean','median','min','max']).reset_index()

df_sous= df_sous.sort_values('mean')
df_sous=df_sous.round(2)
df_sous
```

Afegeix-hi les mitjanes als marges de la taula.

```python
# formula poc exacte perquè fa les mitjanes de les mitjanes.. no les mitjanes reals totals (als marges)
d_sous_pais= df_cap.pivot_table(index='Gènere', columns = "País d'origen", values = 'Salari mensual', aggfunc = 'mean', margins=True, margins_name='Total', fill_value=0)

d_sous_pais=d_sous_pais.round(2)

d_sous_pais.style.background_gradient(cmap='YlOrRd', axis=1)

```

#### 6.
Crea una columna nova que sigui la data de naixament en format Datetime a partir de les columnes dia, mes i any. Utilitzant aquesta columna crea una funció que donada una data, et calculi l'edat actual a dia d'avui.

Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.

```python
df_cap['Data_Naixement'] = pd.to_datetime(df_cap[['Any', 'Mes', 'Dia']].rename(columns={
    'Any': 'year',
    'Mes': 'month',
    'Dia': 'day'
}))

def calcul_edat(naixement):
    if pd.isnull(naixement):
        return None
    avui = date.today()
    edat = avui.year - naixement.year - ((avui.month, avui.day)< (naixement.month, naixement.day))
    return edat

df_cap ['Edat'] = df_cap['Data_Naixement'].apply(calcul_edat)

display(df_cap)
```
