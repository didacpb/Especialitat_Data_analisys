Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.

La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
import numpy as np
from numpy import random 
import string
import pyperclip
from collections import Counter
import pandas as pd
from pprint import (pprint,)
```

```python
def demanar_float(num):
    """
    Demana a l'usuari que introdueixi un valor float i comprova que sigui vàlid

    Arguments: num: El valor introduït per l'usuari

    Retorna: float: El valor introduït per l'usuari
    """
    while True:
       try:
            return float(input('Escriu un valor numèric: '))
            
       except ValueError:
           print("Error: has d'introduïr un valor numèric correcte (utilitza el punt pels decimals)")

def validar_dades (pes, altura):
    """
    Valida que el pes i l'altura estiguin dins dels rangs vàlids

    Arguments: pes / altura

    Retorna: bool: True si les dades són vàlides, False en cas contrari
    """
    return (0 < pes <= 200) and (0 < altura <= 2.40)
    
def demanar_dades():
    """
    Demana a l'usuari que introdueixi el pes i l'altura, i valida les dades

    Retorna: tuple: Una tupla amb el pes i l'altura introduïts per l'usuari
    """
    while True:
        pes = demanar_float('Escriu el teu pes en kilograms: ')
        altura = demanar_float('Escriu la teva altura en metres: ')
        if validar_dades(pes, altura):
            print("Entrada de pes i altura vàlids.")
            return pes, altura
        else:
            print("El pes ha d'estar entre 0 i 200kg i l'altura entre 0 i 2.40m")


def calcul_IMC(pes, altura):
    """
    Calcula l'IMC d'un pes i una altura donada

    Arguments: pes , altura

    Retorna: L'IMC calculat a partir de les dues variables amb dos decimals
    """
    return round(pes/(altura**2),2)

def class_categoria(imc):
    """
    Classifica l'imc en 4 categories dieferents

    Arguments: rep l'imc

    Retorna: la categoria a la qual pertany l'imc
    """
    if imc < 18.5:
        return 'Pes baix'
    elif imc < 25:
        return 'Pes normal'
    elif imc < 30:
        return 'Sobrepès'
    else:
        return 'Obesitat'
    
def mostrar_resultat(imc, categoria):
    """
    Mostra el resultat de l'imc i la categoria corresponent

    Arguments: imc, categoria
    """
    print(f"El teu IMC és: {imc:.2f}; i la categoria és: {categoria}")

def orquestadora_IMC():
    """
    Funció orquestradora que crida a les funcions de càlcul i classificació de l'IMC
    """
    pes,altura = demanar_dades()
    resultatIMC = calcul_IMC(pes, altura)
    categoriaIMC = class_categoria(resultatIMC)
    mostrar_resultat(resultatIMC, categoriaIMC)
```

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

(EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

```python
a_celsius = {
    'C': lambda t: t,
    'K': lambda t: t - 273.15,
    'F': lambda t: (t - 32) * 5/9,
    'Ra': lambda t: (t - 491.67) * 5/9
}

desde_celsius = {
    'C': lambda t: t,
    'K': lambda t: t + 273.15,
    'F': lambda t: (t * 9/5) + 32,
    'Ra': lambda t: (t + 273.15) * 9/5
}

def solicitar_dades():
    while True:
        try:
            val = float(input("Introdueix el valor numèric de la temperatura: "))
            uni = input("En quina unitat està? (C, K, F, Ra): ").upper().strip()
            
            if uni not in a_celsius:
                print(f"Error: Unitat '{uni}' no reconeguda. Intenta-ho amb C, K, F o Ra.")
                continue

            temp_en_c = a_celsius[uni](val)
            if temp_en_c < -273.15:
                print("Error: Temperatura per sota del zero absolut. Impossible!")
                continue
                
            return val, uni
            
        except ValueError:
            print("Error: Introdueix un número vàlid (usa el punt per als decimals).")

def convertidor_universal():
    valor, unitat = solicitar_dades()
    
    temp_c = a_celsius[unitat](valor)
    
    resultats = {u: round(formula(temp_c), 2) for u, formula in desde_celsius.items()}
    
    return resultats

diccionari_final = convertidor_universal()

print("\n--- Conversions Guardades ---")
pprint(diccionari_final)
```

##### 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python
def solicita_text():
    '''Demana un text a l'usuari i el passa a minuscules

    A: text introduit per l'usuari

    R: text en minuscules
    '''
    return input('Escriu un text: ').lower()

def neteja_text(text):
    '''
    Demana un text a l'usuari i el neteja de caràcters que no siguin lletres i espais
    '''     
    text_net= ""   
    for caracter in text:
        if caracter.isalpha() or caracter.isspace():   
            text_net += caracter   
        else:
            text_net += ' '   
    return text_net.split()

def recompte_paraules(llista_paraules):
    '''Conta nº de vegades que apareixen les paraules d'text i les afegeix en un diccionari
    
    A: text net de caracters de la funció solicita_text() i separat amb split()
    
    R: diccionari amb comptador de paraules'''
    
    diccionari_paraules = Counter(llista_paraules)
    return dict(diccionari_paraules)



def mitjana_paraules(dicc):
    ''' Calcula la mitjana de vegades que es repeteixen les paraules en el diccionari creat en la funció recompte_paraules()

    A: diccionari comptador

    R: mitjana de repeticions de paraules'''
    if not dicc: return 0
    valors_series = pd.Series(list(dicc.values()))
    return round(valors_series.mean(), 2)

def mostrar_comptador_paraules(diccionari):
    '''Mostra el comptador de paraules en un format llegible

    A: diccionari comptador

    R: mostra el diccionari comptador'''
    print(diccionari)

def mostrar_mitjana_paraules(mitjana):
    '''Mostra la mitjana de repeticions de paraules en un format llegible

    A: diccionari comptador

    R: mostra la mitjana de repeticions de paraules'''
    
    print('La mitjana de llargada de les paraules del text és de: ', mitjana)

def main_paraules():
    '''
    Funció principal que orquestra les funcions de text
    '''
    text = solicita_text()
    text_net = neteja_text(text)
    dicc_paraules=recompte_paraules(text_net)
    mitjana=mitjana_paraules(dicc_paraules)
    mostrar_comptador_paraules(dicc_paraules)
    mostrar_mitjana_paraules(mitjana)
```

##### 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

Fixa’t en el següents exemples per veure l’output esperat.

```python

def diccionari_girat():
    '''es crea un diccionari a partir d'una enquesta

    A: usuari entre nombre de preguntes de l'enquesta i les preguntes(clau) i respostes(valor)
    
    R: un diccionari amb les clau:valor de l'enquesta'''
    reverse_dicc={}
    for x in range(int(input('Quantes preguntes vols fer?'))):
        reverse_dicc.update({input('Escriu la clau'):input('Escriu el valor')})
    return reverse_dicc



def invertir_diccionari(reverse_dictionary):
    '''
    Rep diccionari i inverteix les claus:valors a valors:claus dipositantles dins un de nou

    A: diccionari

    R: diccionari invertit
    '''
    eldicc={}
    for clau_girat, valor_girat in reverse_dictionary.items(): 
        if valor_girat not in eldicc:
            eldicc[valor_girat]=clau_girat 
        else:
             
            if isinstance(eldicc[valor_girat],list):
                eldicc[valor_girat].append(clau_girat) 
            else:
                valor_antic=eldicc[valor_girat] 
                eldicc[valor_girat]=[valor_antic,clau_girat]
                print(f'La clau {valor_girat} està repetida')  

                
    return eldicc
    

def mostrar_dicc_origional(diccionari):
    '''Mostra el diccionari original en un format llegible

    A: diccionari original

    R: mostra el diccionari original'''
    pprint(f'diccionari_original {diccionari}')

def mostrar_dicc_ordenat(diccionari_ordenat):
    '''Mostra el diccionari invertit en un format llegible

    A: diccionari invertit

    R: mostra el diccionari invertit'''
    pprint(f'diccionari_invertit {diccionari_ordenat}')

def main_diccionari():
    '''
    Funció principal que orquestra les funcions de diccionari
    '''
    diccionari = diccionari_girat()
    mostrar_dicc_origional(diccionari)
    diccionari_invertit = invertir_diccionari(diccionari)
    mostrar_dicc_ordenat(diccionari_invertit)
```

##### 2. Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

```python
llista_exemple = ['1.3','one','1e10','seven','3-1/2',('2',1,1.4,'not-a-number'),[1,2,'3','3.4']]

def separacio_flotants(ll):
    '''funció que rep una llista i converteix els elements en float type

    A: llista introduida per l'usuari

    R: dues llistes amb els valors introduits pero separats per flotants o no'''
    llista_flotant=[]
    llista_resta=[]
    for x in ll:
        if isinstance(x, (list,tuple)):
            for x_llistaotupla in x:
                try:
                    valor_convertit = float(x_llistaotupla)
                    llista_flotant.append(valor_convertit)
                except ValueError:
                    llista_resta.append(x_llistaotupla)
        else:
            try:
                valor_convertit = float(x)
                llista_flotant.append(valor_convertit)
            except ValueError:
                llista_resta.append(x)
    return (llista_resta, llista_flotant)

no_float, si_float=separacio_flotants(llista_exemple)

print(f"Llista original: {llista_exemple}")
print(f"Llista de valors flotants:{si_float}") 
print(f"Llista de valors no flotants: {no_float}")

```

##### 1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)
Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
def seleccio_caracters (majus, minus, num, sign):
    '''funció que genera una llista amb els caracters que composen cada categoria (lletres majuscules, minuscules, numeros del 0 al 9 i signes de puntuacio)
    
    A: rep si els caracters son true o false per inclourels o no a la llista
    
    R: retorna la llista amb els caracters que s'han inclos'''

    caracters_disponibles=""

    if majus:
        caracters_disponibles += string.ascii_uppercase
    if minus:
        caracters_disponibles += string.ascii_lowercase
    if num:
        caracters_disponibles += string.digits
    if sign: 
        caracters_disponibles += string.punctuation
    return list(caracters_disponibles)

def creacio_contrasenya(longitut, majuscules, minuscules, numeros, signes):
    '''funció que crea una contrasenya aleatòria

    A: rep l entrada booleana de cada valor per inclourel a la contrassenya

    R: una cadena aleatòria que conté el nombre i el tipus de caracters seleccionats'''

    llista_caracters= seleccio_caracters(majuscules, minuscules, numeros, signes)
    if not llista_caracters:
        return 'Error: selecciona almenys un tipus de caràcter'
    
    if llista_caracters:
        seleccio= np.random.choice(llista_caracters, size=longitut)
        contrasenya="".join(seleccio)
    return contrasenya

creador_contrasenya=creacio_contrasenya(10, True, True, True, True)


pyperclip.copy(creador_contrasenya)
```

Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

El nombre total de gols que ha fet cada equip.
El nom de l’equip més golejador.
El nom de l’equip més golejat
La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)
Com a referència, els equips amb més punts serien els següents:

import os

```python
#importacio txt a csv
df =pd.read_csv('historic_partits.txt', 
                 sep=r'\s+(?=\d+-\d+)|\b(?<=\d-\d)\s+|\b(?<=\d\d-\d\d)\s+', 
                 header=None, 
                 names=['Local', 'Marcador', 'Visitant'],
                 engine='python')
df[['Gols_L', 'Gols_V']]= df['Marcador'].str.split('-', expand=True).astype(int)

df=df.drop(columns='Marcador')

print(df)
```

```python
#assignem puntuacions
resultats = [
    (df['Gols_L'] > df['Gols_V']),
    (df['Gols_L'] == df['Gols_V']),
    (df['Gols_L'] < df['Gols_V'])
]

punts_l = np.select(resultats, [3,1,0])
punts_v = np.select(resultats, [0,1,3])

df_local = df[['Local','Gols_L', 'Gols_V']].rename(columns={'Local':'Equip', 'Gols_L':'Gols', 'Gols_V':'Gols encaixats'})
df_local['Punts']= punts_l

df_visit = df[['Visitant','Gols_V', 'Gols_L']].rename(columns={'Visitant':'Equip', 'Gols_V':'Gols', 'Gols_L':'Gols encaixats'})
df_visit['Punts']=punts_v


df_local['Partits guanyats'] = (df_local['Punts'] == 3).astype(int)
df_local['Partits empatats'] = (df_local['Punts'] == 1).astype(int)
df_local['Partits perduts'] = (df_local['Punts'] == 0).astype(int)

df_visit['Partits guanyats'] = (df_visit['Punts'] == 3).astype(int)
df_visit['Partits empatats'] = (df_visit['Punts'] == 1).astype(int)
df_visit['Partits perduts'] = (df_visit['Punts'] == 0).astype(int)
```

```python
#creem classificació
df_complert = pd.concat([df_local, df_visit])
df_classif=df_complert.groupby('Equip').agg({'Partits guanyats':'sum', 'Partits empatats':'sum', 'Partits perduts':'sum','Gols':'sum', 'Gols encaixats':'sum', 'Punts':'sum'}).reset_index()

df_classif=df_classif.sort_values(by=['Punts', 'Gols'], ascending=False).reset_index(drop=True)
display(df_classif)
```

```python
maxim_golejador = df_classif.loc[df_classif['Gols'].idxmax(), 'Equip']
print(f"L'equip més golejador és: {maxim_golejador}")

maxim_golejat = df_classif.loc[df_classif['Gols encaixats'].idxmax(), 'Equip']
print(f"L'equip més golejat és: {maxim_golejat}")
```
